## Transformação Box-Tidwell

### Aplicação Prática da Transformação Box-Tidwell

Nesta etapa, vamos aplicar uma transfomação para identificar potências ótimas de transformação para algumas variáveis explicativas.

Tentamos usar a função `boxTidwell` diretamente do pacote `car`, mas não conseguimos executá-la por conta de valores não numéricos. Foi feito uma nova abordagem, dentro de um laço *loop*, na qual identifica as melhores transformações e não precisa remover uma linha inteira.

No fim, decidimos continuar com o par ($x_1$ e $x_2$) a fim de maximizar a relação linear com $y$ (Total de Feridos).

```r
eps <- 1e-3
y <- dados$y
x1p <- dados$x1
x2p <- dados$x2 + eps
x3p <- dados$x3 + eps
x4p <- dados$x4 + eps
x5p <- dados$x5
x6p <- dados$x6
x7p <- dados$x7

bt_result <- list()
vars_bt <- list(
  x1 = x1p, x2 = x2p, x3 = x3p, x4 = x4p,
  x5 = x5p, x6 = x6p, x7 = x7p
)

for (nm in names(vars_bt)) {
  res <- tryCatch(
    boxTidwell(y ~ vars_bt[[nm]]),
    error = function(e) {cat("ERRO:", conditionMessage(e), " Não convergiu. \n"); NULL}
  )
  if (!is.null(res)) {
    print(res)
    bt_result[[nm]] <- res
  }
}
```

O Box–Tidwell não convergiu para populacao (há municípios de escala muito diferente — Belo Horizonte ≈ 2,4 milhões contra < 3 mil em Senador José Bento), mas o gráfico de dispersão deixa claro o perfil logarítmico. Para `prop_urbano` o $λ$ estimado é $≈ 1,5$, mas como as correlações de $x_{2}$ e $√x_{2}$ são muito próximas, a transformação é pouco relevante.

Para `prop_chuva` o algoritmo também não convergiu adequadamente (atingiu o número máximo de iterações = 26), produzindo $λ ≈ -9,72$, que não tem interpretação prática (descartado). Para `prop_noite` e `media_veiculos_por_acidente` o Box-Tidwell também não convergiu (NA/NaN/Inf), indicando que essas variáveis não têm forma funcional estável que linearize sua relação com y.

## Parte 2

Um modelo de regressão linear simples relaciona a variável resposta $Y$ com apenas uma variável explicativa $x$

\
$$Y_{i}|x_{i}=\beta_{0}+\beta_{1}x_{i}+E_{i},$$

\
sendo $E_{i}$~$N(0,\sigma^{2})$ e $cov(E_{i},E_{j})=0$ para $i\ne j$.

\
Assim sendo, tem-se

$$\mu_{i}=E(Y_{i}|x_{i})=\beta_{0}+\beta_{1}x_{i}$$

## Tipos de resíduos

Os resíduos são utilizados na verificação das suposições do modelo de regressão e na identificação de observações atípicas. Existem diferentes tipos de resíduos, sendo alguns mais apropriados para a avaliação das suposições e outros para a identificação de observações atípicas.

Esses resíduos se distinguem pela forma como são calculados, especialmente pela maneira como consideram a variabilidade dos erros e a alavancagem das observações.

### Resíduos ordinários

Os resíduos ordinários são definidos como as diferenças entre os valores observados e os valores ajustados pelo modelo de regressão.

Sua expressão é:

$$
e_i = y_i - \hat{y}_i
$$

em que:

- $y_i$ é o valor observado da variável resposta;
- $\hat{y}_i$ é o valor ajustado pelo modelo.

No R, são calculados pela função `resid(modelo)`.

```r
modelo <- lm(y~x)
residuos<-resid(modelo)
residuos
```

### Resíduos padronizados

Os resíduos padronizados são obtidos dividindo cada resíduo ordinário pela estimativa do desvio-padrão dos erros.

Sua expressão é:

$$
z_i = \frac{e_i}{\hat{\sigma}}
    = \frac{y_i-\hat{y}_i}{\sqrt{QM_{\text{Erro}}}}
$$

em que:

- $e_i=y_i-\hat{y}_i$ é o resíduo ordinário;
- $\hat{\sigma}^2=QM_{\text{Erro}}$ é a estimativa da variância dos erros;
- $\hat{\sigma}=\sqrt{QM_{\text{Erro}}}$ é a estimativa do desvio-padrão dos erros.

No R, os resíduos padronizados podem ser calculados por:

`resid(modelo) / sigma(modelo)`

```r
resid(modelo) / sigma(modelo)

estimativa_ep_erro=sqrt(anova(modelo)[["Mean Sq"]][2]);  estimativa_ep_erro   # estimativa do erro padrão dos erros

res_padronizado=residuos/estimativa_ep_erro; res_padronizado # resíduos padronizados
```

### Resíduo studentizado internamente

O resíduo studentizado internamente é calculado pela função `rstandard(modelo)` no R.

Sua expressão é:

$$
r_i = \frac{e_i}{\hat{\sigma}\sqrt{1-h_{ii}}}
$$

em que:

- $e_i = y_i - \hat{y}_i$ é o resíduo ordinário;
- $\hat{\sigma} = \sqrt{QM_{\text{Resíduo}}}$ é a estimativa do desvio-padrão dos erros;
- $h_{ii}$ é o elemento da diagonal principal da matriz de projeção:

$$
H = X(X'X)^{-1}X'
$$

O resíduo studentizado internamente também é denominado **resíduo padronizado internamente**.

```r
res_pad<-rstandard(modelo); res_pad
```

### Resíduos studentizados externamente (utilizado na análise de influencia)

Os resíduos studentizados externamente são obtidos dividindo cada resíduo ordinário por uma estimativa de seu desvio-padrão, calculada sem considerar a observação correspondente. Esses resíduos são especialmente úteis na identificação de observações atípicas.

Sua expressão é:

$$
t_i = \frac{e_i}{\hat{\sigma}_{(i)}\sqrt{1-h_{ii}}}
$$

em que:

- $e_i=y_i-\hat{y}_i$ é o resíduo ordinário;
- $\hat{\sigma}_{(i)}$ é a estimativa do desvio-padrão dos erros, obtida sem a observação $i$;
- $h_{ii}$ é o elemento da diagonal principal da matriz de projeção $H=X(X'X)^{-1}X'$.

No R, são calculados pela função `rstudent(modelo)`.

```r
r_studentizado <- rstudent(modelo)
r_studentizado
```

```r
# testes de normalidade
t_norm_shapiro <- shapiro.test(rstandard(modelo))
t_norm_ad <- nortest::ad.test(rstandard(modelo))
t_norm_lillie <- nortest::lillie.test(rstandard(modelo))
t_norm_cvm <- nortest::cvm.test(rstandard(modelo))
t_norm_sf <- nortest::sf.test(rstandard(modelo))
alpha_norm <- 0.05

testes_norm <- c(
  "Shapiro–Wilk", "Anderson–Darling", "Lilliefors",
  "Cramér–von Mises", "Shapiro–Francia"
)
estatisticas_norm <- as.numeric(c(
  t_norm_shapiro$statistic, t_norm_ad$statistic,
  t_norm_lillie$statistic, t_norm_cvm$statistic,
  t_norm_sf$statistic
))
pvalores_norm <- c(
  t_norm_shapiro$p.value, t_norm_ad$p.value,
  t_norm_lillie$p.value, t_norm_cvm$p.value,
  t_norm_sf$p.value
)
resultados_norm <- data.frame(
  Teste = testes_norm,
  Estatistica = estatisticas_norm,
  p_valor = pvalores_norm,
  Decisao = ifelse(pvalores_norm <= alpha_norm,
                   "Rejeitar H0", "Não rejeitar H0")
)

# Manter os valores numéricos originais e formatar apenas a exibição.
tabela_norm <- resultados_norm
tabela_norm$Estatistica <- formatC(estatisticas_norm,
                                  format = "f", digits = 4, decimal.mark = ",")
tabela_norm$p_valor <- ifelse(
  pvalores_norm < 1e-12,
  "< 0,000000000001",
  formatC(pvalores_norm, format = "f", digits = 12, decimal.mark = ",")
)
knitr::kable(tabela_norm,
             col.names = c("Teste", "Estatística", "p-valor", "Decisão"),
             row.names = FALSE)

```

## 2. Análise de regressão polinomial de 2º grau

Verificamos, individualmente, se cada variável explicativa tem relação quadrática com a resposta, ajustando para cada $x_i$ o modelo

$$\log(Y_i) = \beta_0 + \beta_1 x_i + \beta_2 x_i^2 + E_i,$$

com `lm(ly ~ xi + I(xi^2))`.

- **2a)** Verificamos as suposições de cada modelo (normalidade, homogeneidade de variâncias, independência e outliers) com os mesmos procedimentos da 1ª entrega.
- **2b)** Avaliamos se o termo quadrático é apropriado pelo teste $H_0: \beta_2 = 0$ vs $H_a: \beta_2 \ne 0$ (p-valor do coeficiente `I(xi^2)` no `summary`) junto com a análise das suposições.

Para não repetir o mesmo código para cada variável, a função abaixo reúne a análise gráfica dos resíduos e os testes usados na 1ª entrega.

```r
# Função de diagnóstico: reúne a verificação de suposições da 1ª entrega
diagnostico <- function(mod, nome = "modelo", graficos = TRUE, alpha = 0.05) {
  r_int <- rstandard(mod)   # resíduos studentizados internamente
  pred  <- fitted(mod)

  if (graficos) {
    par(mfrow = c(2, 2), mar = c(4, 4, 2, 1), oma = c(0, 0, 2, 0))
    boxplot(r_int, ylab = "resíduos studentizados", main = "Boxplot dos resíduos")
    points(mean(r_int), col = "red", pch = 3)
    car::qqPlot(mod, envelope = 0.95, ylab = "resíduos studentizados",
                main = "QQ-plot com envelope 95%", id = FALSE)
    plot(pred, r_int, xlab = "preditos", ylab = "resíduos studentizados",
         main = "Resíduos x preditos", pch = 19, col = rgb(0, 0, 1, 0.6)); abline(h = 0)
    hist(r_int, main = "Histograma dos resíduos", xlab = "resíduos studentizados",
         col = "orange", breaks = 20)
    mtext(nome, outer = TRUE, cex = 1.1, font = 2)
    par(mfrow = c(1, 1), oma = c(0, 0, 0, 0))
  }

  sw <- shapiro.test(r_int)
  ad <- nortest::ad.test(r_int)
  li <- nortest::lillie.test(r_int)
  bp <- lmtest::bptest(mod)
  nc <- car::ncvTest(mod)
  dw <- car::durbinWatsonTest(mod)
  ot <- car::outlierTest(mod)

  pv <- c(sw$p.value, ad$p.value, li$p.value, bp$p.value, nc$p, dw$p)
  data.frame(
    Modelo   = nome,
    Suposicao = c("Normalidade", "Normalidade", "Normalidade",
                  "Homocedasticidade", "Homocedasticidade", "Independência"),
    Teste    = c("Shapiro-Wilk", "Anderson-Darling", "Lilliefors",
                 "Breusch-Pagan", "ncvTest", "Durbin-Watson"),
    p_valor  = signif(pv, 4),
    Decisao  = ifelse(pv <= alpha, "Rejeitar H0", "Não rejeitar H0"),
    Outliers_Bonferroni = sum(ot$bonf.p < alpha),
    row.names = NULL
  )
}
```

### 2.1) Modelo quadrático em $log(x_1)$ — log(população)

```r
modelo_quad_lx1 <- lm(ly ~ lx1 + I(lx1^2), data = dm)
summary(modelo_quad_lx1)

# curva ajustada sobre o diagrama de dispersão
par(mfrow = c(1, 1), mar = c(5, 5, 2, 1))
plot(ly ~ lx1, data = dm, pch = 19, col = rgb(0, 0, 1, 0.5),
     xlab = "lx1", ylab = "log(y)", main = "Ajuste quadrático: log(y) ~ lx1 + lx1^2")
grade <- data.frame(lx1 = seq(min(dm$lx1), max(dm$lx1), length.out = 200))
lines(grade$lx1, predict(modelo_quad_lx1, grade), col = "red", lwd = 2)
abline(lm(ly ~ lx1, data = dm), col = "darkgreen", lty = 2, lwd = 2)
legend("bottomright", c("quadrático", "linear"), col = c("red", "darkgreen"),
       lty = c(1, 2), lwd = 2, bty = "n")

# 2a) suposições
knitr::kable(diagnostico(modelo_quad_lx1, "log(y) ~ lx1 + lx1^2"))
```

**2b) O modelo quadrático foi apropriado? Não.** O termo quadrático não é significativo ($\hat\beta_2 = 0{,}0138$, p-valor $= 0{,}562$) e o $R^2$ ajustado cai de $0{,}4155$ (modelo linear da 1ª entrega) para $0{,}4135$. As suposições continuam com os mesmos problemas do modelo linear: normalidade rejeitada pelos três testes, por causa da cauda inferior, e homocedasticidade rejeitada pelo `ncvTest` ($p = 0{,}010$), mas não pelo Breusch-Pagan ($p = 0{,}150$). A relação entre $\log(y)$ e $\log(x_1)$ é linear, e mantemos o termo de 1º grau.

Por isso, parte do efeito quadrático pode ser um artefato do tamanho do município, e não um efeito real do período noturno.

### 2.4) Modelo quadrático em $x_4$ — proporção de acidentes com chuva/chuvisco

```r
modelo_quad_x4 <- lm(ly ~ x4 + I(x4^2), data = dm)
summary(modelo_quad_x4)

# curva ajustada sobre o diagrama de dispersão
par(mfrow = c(1, 1), mar = c(5, 5, 2, 1))
plot(ly ~ x4, data = dm, pch = 19, col = rgb(0, 0, 1, 0.5),
     xlab = "x4", ylab = "log(y)", main = "Ajuste quadrático: log(y) ~ x4 + x4^2")
grade <- data.frame(x4 = seq(min(dm$x4), max(dm$x4), length.out = 200))
lines(grade$x4, predict(modelo_quad_x4, grade), col = "red", lwd = 2)
abline(lm(ly ~ x4, data = dm), col = "darkgreen", lty = 2, lwd = 2)
legend("bottomright", c("quadrático", "linear"), col = c("red", "darkgreen"),
       lty = c(1, 2), lwd = 2, bty = "n")

# 2a) suposições
knitr::kable(diagnostico(modelo_quad_x4, "log(y) ~ x4 + x4^2"))
```

**2b) O modelo quadrático foi apropriado? Estatisticamente sim, mas o ganho é pequeno.** O termo quadrático é significativo ($\hat\beta_2 = -9{,}64$, p-valor $= 0{,}005$), e as suposições de normalidade e homocedasticidade são atendidas. Porém, o modelo explica muito pouco: o $R^2$ ajustado vai de $-0{,}002$ para apenas $0{,}034$. O máximo da curva fica em $x_4 \approx 0{,}23$. O mesmo cuidado do item anterior vale aqui: proporções extremas de chuva ocorrem principalmente em municípios com poucos acidentes.

### 2.5) Modelo quadrático em $x_7$ — média de veículos por acidente

```r
modelo_quad_x7 <- lm(ly ~ x7 + I(x7^2), data = dm)
summary(modelo_quad_x7)

# curva ajustada sobre o diagrama de dispersão
par(mfrow = c(1, 1), mar = c(5, 5, 2, 1))
plot(ly ~ x7, data = dm, pch = 19, col = rgb(0, 0, 1, 0.5),
     xlab = "x7", ylab = "log(y)", main = "Ajuste quadrático: log(y) ~ x7 + x7^2")
grade <- data.frame(x7 = seq(min(dm$x7), max(dm$x7), length.out = 200))
lines(grade$x7, predict(modelo_quad_x7, grade), col = "red", lwd = 2)
abline(lm(ly ~ x7, data = dm), col = "darkgreen", lty = 2, lwd = 2)
legend("bottomright", c("quadrático", "linear"), col = c("red", "darkgreen"),
       lty = c(1, 2), lwd = 2, bty = "n")

# 2a) suposições
knitr::kable(diagnostico(modelo_quad_x7, "log(y) ~ x7 + x7^2"))
```

**2b) O modelo quadrático foi apropriado? Estatisticamente sim, mas com pouco poder explicativo.** O termo quadrático é significativo ($\hat\beta_2 = -1{,}14$, p-valor $= 0{,}001$), e as suposições de normalidade e homocedasticidade são atendidas. Ainda assim, o $R^2$ ajustado é de apenas $0{,}043$. O máximo da curva fica em $x_7 \approx 2{,}18$ veículos por acidente.

*Observação sobre independência (todos os modelos):* o Durbin-Watson rejeita $H_0$ em todos os ajustes porque o arquivo está ordenado de forma decrescente por `total_feridos`. Para dados de corte transversal esse teste depende da ordem arbitrária das linhas; isso é verificado na seção 3.

### Resumo dos modelos quadráticos

A tabela abaixo reúne, para cada variável, o p-valor do coeficiente quadrático e o $R^2$ ajustado dos modelos linear e quadrático.

```r
vars_quad <- c("lx1", "x2", "x3", "x4", "x7")
resumo_quad <- do.call(rbind, lapply(vars_quad, function(v) {
  m_lin  <- lm(as.formula(paste0("ly ~ ", v)), data = dm)
  m_quad <- lm(as.formula(paste0("ly ~ ", v, " + I(", v, "^2)")), data = dm)
  cf <- summary(m_quad)$coefficients
  data.frame(
    Variavel        = v,
    beta2           = signif(cf[3, 1], 4),
    p_valor_beta2   = signif(cf[3, 4], 4),
    R2aj_linear     = round(summary(m_lin)$adj.r.squared, 4),
    R2aj_quadratico = round(summary(m_quad)$adj.r.squared, 4),
    Termo_quadratico = ifelse(cf[3, 4] <= 0.05, "significativo", "não significativo")
  )
}))
knitr::kable(resumo_quad, row.names = FALSE)
```

**Conclusão do item 2:**

* $\log(x_1)$ e $x_2$: o termo quadrático **não** é significativo, e as relações com $\log(y)$ são lineares.
* $x_3$, $x_4$ e $x_7$: o termo quadrático é significativo (relação em "U invertido"). Mas, exceto para $x_3$, o ganho em $R^2$ ajustado é pequeno, e parte do efeito pode vir dos municípios com poucos acidentes, onde as proporções são instáveis.
* Seguindo o roteiro, a seleção de variáveis (item 3) parte do modelo completo com os termos de 1º grau. Ao final do item 3 verificamos, como análise complementar, se os termos quadráticos significativos acrescentam algo ao modelo selecionado.
